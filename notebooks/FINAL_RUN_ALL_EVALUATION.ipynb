```python
from pathlib import Path
import os
import subprocess
import sys

PROJECT_ROOT = Path(r"C:\Users\cdipi\Downloads\aqi_for india")
NOTEBOOK_DIR = PROJECT_ROOT / "notebooks"
EXECUTED_DIR = PROJECT_ROOT / "reports" / "executed_notebooks"
EXECUTED_DIR.mkdir(parents=True, exist_ok=True)

# STEP_1 includes the original benchmark/model notebook, so RUN_THIS_AQI_RESEARCH need not run separately.
PIPELINE = [
    NOTEBOOK_DIR / "STEP_1_BASELINES_AND_EVALUATION.ipynb",
    NOTEBOOK_DIR / "STEP_2_CALIBRATION_AND_ALERTS.ipynb",
    NOTEBOOK_DIR / "STEP_3_ROBUSTNESS_AND_CONFIDENCE_INTERVALS.ipynb",
]
missing = [str(path) for path in PIPELINE if not path.exists()]
assert not missing, f"Missing supporting notebooks: {missing}"
print("Ready to run:")
for item in PIPELINE:
    print(" -", item.name)
```

## Run the full pipeline

Run the next cell once. It stops immediately if any stage fails, so the error remains visible.


```python
for notebook_path in PIPELINE:
    print(f"\n{'=' * 70}\nRunning {notebook_path.name}\n{'=' * 70}")
    command = [
        sys.executable, "-m", "jupyter", "nbconvert",
        "--to", "notebook", "--execute", str(notebook_path),
        "--output-dir", str(EXECUTED_DIR),
        "--ExecutePreprocessor.timeout=1800",
    ]
    subprocess.run(command, cwd=PROJECT_ROOT, check=True, env={**os.environ, "MPLBACKEND": "Agg"})

print("\nComplete ? all final outputs were generated.")
```
